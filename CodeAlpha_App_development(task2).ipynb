```python
# Install ipywidgets
!pip install ipywidgets -q

import random
import html
import ipywidgets as widgets
from IPython.display import display
```

```python
quotes = [
    {
        "text": "The only way to do great work is to love what you do.",
        "author": "Steve Jobs"
    },
    {
        "text": "It always seems impossible until it's done.",
        "author": "Nelson Mandela"
    },
    {
        "text": "Success is not final, failure is not fatal.",
        "author": "Winston Churchill"
    },
    {
        "text": "Believe you can and you're halfway there.",
        "author": "Theodore Roosevelt"
    },
    {
        "text": "The future depends on what you do today.",
        "author": "Mahatma Gandhi"
    },
    {
        "text": "Don't watch the clock; do what it does. Keep going.",
        "author": "Sam Levenson"
    },
    {
        "text": "The secret of getting ahead is getting started.",
        "author": "Mark Twain"
    },
    {
        "text": "It does not matter how slowly you go as long as you do not stop.",
        "author": "Confucius"
    },
    {
        "text": "The best way to predict the future is to create it.",
        "author": "Peter Drucker"
    },
    {
        "text": "Great things are done by a series of small things brought together.",
        "author": "Vincent van Gogh"
    }
]
```

```python
last_index = None
```

```python
quote_display = widgets.HTML()

def show_random_quote():
    global last_index

    # Prevent the same quote from appearing twice
    available = [
        i for i in range(len(quotes))
        if i != last_index
    ]

    index = random.choice(available)
    last_index = index

    quote = quotes[index]

    quote_text = html.escape(quote["text"])
    author = html.escape(quote["author"])

    quote_display.value = f"""
    <div style="
        width: 600px;
        max-width: 90%;
        margin: 20px auto;
        padding: 40px;
        text-align: center;
        border-radius: 15px;
        background: #f7f7f7;
        border: 1px solid #dddddd;
    ">

        <div style="
            font-size: 45px;
            margin-bottom: 15px;
        ">❝</div>

        <div style="
            font-size: 24px;
            line-height: 1.5;
            font-style: italic;
            margin-bottom: 25px;
        ">
            {quote_text}
        </div>

        <div style="
            font-size: 18px;
            font-weight: bold;
        ">
            — {author}
        </div>

    </div>
    """
```

```python
new_quote_button = widgets.Button(
    description=" New Quote",
    button_style="primary",
    layout=widgets.Layout(
        width="150px",
        height="40px"
    )
)

def new_quote(button):
    show_random_quote()

new_quote_button.on_click(new_quote)
```

```python
title = widgets.HTML(
    """
    <div style="text-align:center; margin-bottom:20px;">
        <h1>💬 Random Quote Generator</h1>
        <p style="font-size:16px; color:#666;">
            Get inspired with a new quote!
        </p>
    </div>
    """
)
```

```python
button_box = widgets.HBox(
    [new_quote_button],
    layout=widgets.Layout(
        justify_content="center"
    )
)
```

```python
app = widgets.VBox(
    [
        title,
        quote_display,
        button_box
    ],
    layout=widgets.Layout(
        align_items="center",
        width="100%"
    )
)

display(app)
show_random_quote()
```
