```python
selected_language = "Spanish"
selected_category = "Vocabulary"
current_card = 0
show_translation = False

quiz_questions = []
quiz_index = 0
quiz_score = 0

progress = {
    "Spanish": 0,
    "French": 0,
    "German": 0
}
```

```python
progress_file = "language_progress.json"

def save_progress():
    with open(progress_file, "w") as file:
        json.dump(progress, file)

def load_progress():
    global progress

    if os.path.exists(progress_file):
        try:
            with open(progress_file, "r") as file:
                progress = json.load(file)
        except:
            pass

load_progress()
```

```python
title = widgets.HTML(
    """
    <div style="text-align:center;">
        <h1>🌎 Language Learning App</h1>
        <p>Learn vocabulary, grammar and practice your skills.</p>
    </div>
    """
)

language_dropdown = widgets.Dropdown(
    options=["Spanish", "French", "German"],
    value="Spanish",
    description="Language:"
)

category_dropdown = widgets.Dropdown(
    options=["Vocabulary", "Grammar"],
    value="Vocabulary",
    description="Category:"
)

card_display = widgets.HTML()

status_display = widgets.HTML()

translation_button = widgets.Button(
    description="Show Translation",
    button_style="info"
)

previous_button = widgets.Button(
    description="⬅ Previous"
)

next_button = widgets.Button(
    description="Next ➡"
)

quiz_button = widgets.Button(
    description="📝 Start Quiz",
    button_style="success"
)

progress_display = widgets.HTML()

quiz_output = widgets.Output()
```

```python
def display_card():
    global show_translation

    data = language_data[selected_language][selected_category]

    if not data:
        card_display.value = "<h3>No lessons available.</h3>"
        return

    card = data[current_card]

    word = card["word"]
    pronunciation = card["pronunciation"]

    if show_translation:
        translation = card["translation"]

        card_display.value = f"""
        <div style="
            padding:35px;
            margin:20px;
            text-align:center;
            border:1px solid #ddd;
            border-radius:15px;
            background:#f8f9fa;
        ">
            <h2>{word}</h2>

            <p style="font-size:18px;">
                🔊 Pronunciation: <b>{pronunciation}</b>
            </p>

            <hr>

            <h3>Translation</h3>

            <p style="font-size:24px;">
                {translation}
            </p>
        </div>
        """

        translation_button.description = "Hide Translation"

    else:

        card_display.value = f"""
        <div style="
            padding:35px;
            margin:20px;
            text-align:center;
            border:1px solid #ddd;
            border-radius:15px;
            background:#f8f9fa;
        ">
            <h2>{word}</h2>

            <p style="font-size:18px;">
                🔊 Pronunciation: <b>{pronunciation}</b>
            </p>

            <p style="color:#777;">
                Click "Show Translation" to reveal the meaning.
            </p>
        </div>
        """

        translation_button.description = "Show Translation"

    status_display.value = (
        f"<p style='text-align:center;'>"
        f"Card {current_card + 1} of {len(data)}"
        f"</p>"
    )
```

```python
def change_language(change):
    global selected_language
    global current_card
    global show_translation

    selected_language = change["new"]

    current_card = 0
    show_translation = False

    display_card()
    display_progress()

language_dropdown.observe(change_language, names="value")
```

```python
def change_category(change):
    global selected_category
    global current_card
    global show_translation

    selected_category = change["new"]

    current_card = 0
    show_translation = False

    display_card()

category_dropdown.observe(change_category, names="value")
```

```python
def toggle_translation(button):
    global show_translation

    show_translation = not show_translation

    display_card()

translation_button.on_click(toggle_translation)
```

```python
def next_card_function(button):
    global current_card
    global show_translation

    data = language_data[selected_language][selected_category]

    current_card = (current_card + 1) % len(data)

    show_translation = False

    display_card()

next_button.on_click(next_card_function)
```

```python
def previous_card_function(button):
    global current_card
    global show_translation

    data = language_data[selected_language][selected_category]

    current_card = (current_card - 1) % len(data)

    show_translation = False

    display_card()

previous_button.on_click(previous_card_function)
```

```python
def start_quiz(button):
    global quiz_questions
    global quiz_index
    global quiz_score

    data = language_data[selected_language][selected_category]

    quiz_questions = random.sample(
        data,
        min(5, len(data))
    )

    quiz_index = 0
    quiz_score = 0

    quiz_output.clear_output()

    show_quiz_question()
```

```python
def show_quiz_question():

    with quiz_output:
        clear_output()

        if quiz_index >= len(quiz_questions):

            progress[selected_language] += quiz_score
            save_progress()

            print("=" * 40)
            print("🎉 QUIZ COMPLETED!")
            print("=" * 40)
            print()
            print(
                f"Score: {quiz_score}/{len(quiz_questions)}"
            )
            print()

            display_progress()

            return

        question = quiz_questions[quiz_index]

        correct_answer = question["translation"]

        # Create incorrect answers
        all_answers = [
            item["translation"]
            for item in language_data[selected_language]
            ["Vocabulary"]
        ]

        all_answers = list(set(all_answers))

        incorrect = [
            answer
            for answer in all_answers
            if answer != correct_answer
        ]

        random.shuffle(incorrect)

        options = incorrect[:3] + [correct_answer]

        random.shuffle(options)

        print(
            f"Question {quiz_index + 1} "
            f"of {len(quiz_questions)}"
        )

        print()

        print(
            f"What does '{question['word']}' mean?"
        )

        print()

        buttons = []

        for option in options:

            button = widgets.Button(
                description=option,
                layout=widgets.Layout(
                    width="300px"
                )
            )

            button.on_click(
                lambda b, selected=option:
                check_answer(selected)
            )

            buttons.append(button)
            display(button)
```

```python
def check_answer(selected_answer):

    global quiz_index
    global quiz_score

    correct_answer = quiz_questions[
        quiz_index
    ]["translation"]

    with quiz_output:
        if selected_answer == correct_answer:
            print("\n✅ Correct!")
            quiz_score += 1
        else:
            print(
                f"\n❌ Incorrect!"
                f"\nCorrect answer: {correct_answer}"
            )

    quiz_index += 1

    # Continue to next question
    show_quiz_question()

quiz_button.on_click(start_quiz)

# ==========================================================
# PROGRESS
# ==========================================================

def display_progress():

    score = progress[selected_language]

    progress_display.value = f"""
    <div style="
        padding:15px;
        margin-top:15px;
        text-align:center;
        border-radius:10px;
        background:#eef6ff;
    ">
         <b>{selected_language} Progress:</b>
        {score} quiz points
    </div>
    """
```

```python
controls = widgets.HBox(
    [
        language_dropdown,
        category_dropdown
    ],
    layout=widgets.Layout(
        justify_content="center"
    )
)

navigation = widgets.HBox(
    [
        previous_button,
        translation_button,
        next_button
    ],
    layout=widgets.Layout(
        justify_content="center"
    )
)

main_ui = widgets.VBox(
    [
        title,
        controls,
        card_display,
        status_display,
        navigation,
        progress_display,
        quiz_button,
        quiz_output
    ],
    layout=widgets.Layout(
        width="80%",
        margin="auto"
    )
)

display(main_ui)

# Display first card
display_card()
display_progress()
```
