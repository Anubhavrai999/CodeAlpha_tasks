```python
import gradio as gr
from deep_translator import GoogleTranslator
from gtts import gTTS
import os
```

```python
languages = {
    "Auto Detect": "auto",
    "English": "en",
    "Hindi": "hi",
    "Bengali": "bn",
    "Tamil": "ta",
    "Telugu": "te",
    "Malayalam": "ml",
    "Kannada": "kn",
    "Gujarati": "gu",
    "Punjabi": "pa",
    "Marathi": "mr",
    "Urdu": "ur",
    "French": "fr",
    "German": "de",
    "Spanish": "es",
    "Italian": "it",
    "Chinese": "zh-CN",
    "Japanese": "ja",
    "Korean": "ko",
    "Russian": "ru",
    "Arabic": "ar"
}
```

```python
def translate_text(text, source, target):
    if text.strip() == "":
        return "Please enter some text."

    try:
        translated = GoogleTranslator(
            source=languages[source],
            target=languages[target]
        ).translate(text)

        return translated

    except Exception as e:
        return f"Error: {str(e)}"
```

```python
def text_to_speech(text):
    if text.strip() == "":
        return None

    tts = gTTS(text=text)
    filename = "translated_audio.mp3"
    tts.save(filename)
    return filename
```

```python
with gr.Blocks(title="Language Translation Tool") as demo:

    gr.Markdown("#  Language Translation Tool")
    gr.Markdown("Translate text between multiple languages using Google Translate API.")

    input_text = gr.Textbox(
        lines=6,
        label="Enter Text"
    )

    with gr.Row():

        source_lang = gr.Dropdown(
            choices=list(languages.keys()),
            value="Auto Detect",
            label="Source Language"
        )

        target_lang = gr.Dropdown(
            choices=list(languages.keys()),
            value="Hindi",
            label="Target Language"
        )

    translate_btn = gr.Button("Translate")

    output_text = gr.Textbox(
        lines=6,
        label="Translated Text"
    )

    copy_btn = gr.Button("Copy Translation")

    copy_status = gr.Textbox(
        label="Status",
        interactive=False
    )

    audio_output = gr.Audio(label="Text-to-Speech")

    translate_btn.click(
        fn=translate_text,
        inputs=[input_text, source_lang, target_lang],
        outputs=output_text
    )

    output_text.change(
        fn=text_to_speech,
        inputs=output_text,
        outputs=audio_output
    )

    copy_btn.click(
        lambda x: "✅ Copy the translated text manually (Ctrl+C).",
        inputs=output_text,
        outputs=copy_status
    )

demo.launch()
```
